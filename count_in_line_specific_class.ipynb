```python
import cv2
from ultralytics import solutions
import time
```

```python
# cap = cv2.VideoCapture(r"videos\traffic_camera_ongoing.mp4")
cap = cv2.VideoCapture(r"videos\vehicle_counting_2.mp4")
assert cap.isOpened(), "Error reading video file"
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
```

```python
# Video writer
video_writer = cv2.VideoWriter("object_counting_output.avi", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
```

```python
# Define line points
line_points = [(21, 540), (1234, 540)]
# line_points = [(158, 624), (486, 624)] # for vehicle ongoing video
```

```python
# Init Object Counter
counter = solutions.ObjectCounter(
    show=False,
    region=line_points,
    model=r"models\yolo11m.pt",
    classes=[2, 5, 7],
)
```

```python
# Record the start time
start_time = time.time()

# Process video
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        print("Video frame is empty or video processing has been successfully completed.")
        break
    im0 = counter.count(im0)
    video_writer.write(im0)

# Record the end time
end_time = time.time()
# Calculate total processing time
total_processing_time = end_time - start_time


cap.release()
video_writer.release()
cv2.destroyAllWindows()

# Print total processing time
print(f"Total processing time: {total_processing_time:.2f} seconds")
```
