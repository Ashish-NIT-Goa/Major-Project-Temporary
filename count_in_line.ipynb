```python
import cv2
from ultralytics import YOLO, solutions
```

```python
model = YOLO(r"models\yolo11m.pt")
```

```python
cap = cv2.VideoCapture(r"videos\vehicle_counting_2.mp4")
assert cap.isOpened(), "Error reading video file"
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
```

```python
# set new w and h with smaller resolution and with same aspect ratio. don't use solutions
# w, h = w*2//3, h*2//3
```

```python
# Define line points
line_points = [(21, 540), (1234, 540)]
```

```python
# Video writer
video_writer = cv2.VideoWriter("object_counting_output.avi", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
```

```python
# Init Object Counter
counter = solutions.ObjectCounter(
    view_img=True,
    reg_pts=line_points,
    names=model.names, # here we can change the dictionary of the names
    draw_tracks=False,
    line_thickness=2,
)
```

```python
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        print("Video frame is empty or video processing has been successfully completed.")
        break
    tracks = model.track(im0, persist=True, show=False)
    im0 = counter.start_counting(im0, tracks)
    # video_writer.write(im0)

cap.release()
# video_writer.release()
cv2.destroyAllWindows()
```
